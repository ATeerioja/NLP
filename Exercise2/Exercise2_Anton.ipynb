```python
import json
import tqdm
import math
from transformers import AutoTokenizer
from collections import defaultdict, Counter
```

```python
tokenizer = AutoTokenizer.from_pretrained("distilbert/distilbert-base-uncased")

with open("imdb_train.json", "rt", encoding="utf-8") as f:
    corpus = json.load(f)

tokenized_corpus = [] # [[doc1],[doc2],[doc3],...]

for doc in tqdm.tqdm(corpus[1000:2000]): # IMDB documents
    tokenized = tokenizer.tokenize(doc["text"])
    tokenized_corpus.append(tokenized)
    
```

```python
N = len(tokenized_corpus)
doc_frequency = defaultdict(int) # [('token', 'count of appearances in unique docs aka df')]

for token_ids in (tokenized_corpus):
  for token_id in set(token_ids):
    doc_frequency[token_id] += 1
```

```python
idf = {
  token_id: math.log((1 + N) / (1 + df)) + 1
  for token_id, df in doc_frequency.items()
}

sorted_idf = sorted(idf.items(), key=lambda x: x[1], reverse=True)

top_10 = sorted_idf[:10]
bottom_10 = sorted_idf[-10:]

print("TOP 10")
for token_id, score in top_10:
    print(f"{token_id:<15} -> {score}")

print("\nBOTTOM 10")
for token_id, score in bottom_10:
    print(f"{token_id:<15} -> {score}")
```

```python
def tf_idf(doc):
  tokenized_doc = tokenizer.tokenize(doc)
  token_counter = Counter()
  token_counter.update(tokenized_doc)
  
  tf_idf = [] # [('token','tf_idf value of token')]

  for item in token_counter.items():
    tf_idf.append((item[0], (item[1] * idf[item[0]])))

  tf_idf = sorted(tf_idf, key=lambda x: x[1], reverse=True)
  top_10 = tf_idf[:10]

  print("TOP 10")
  for i in top_10:
    print(f"{i[0]:<15} -> {i[1]}")

tf_idf(corpus[1002]['text'])  
```
