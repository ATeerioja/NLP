```python
with open("enwiki-20220301-sample-tokenized.txt", mode="rt", encoding='utf-8') as f:
  data = f.read().split("\n")
```

```python
dataset = []

for line in data:
  dataset.append(line.split(" "))
```

```python
from gensim.models import Word2Vec

# train model
model = Word2Vec(
  dataset,
  sg=0,
  vector_size=40,
  window=2,
  min_count=1,
  workers=6,
  epochs=10
)
```

# PART A

```python
for context in [['Volvo'],['Audi'],['Nissan']]:
    print("Context:", context)
    pred = model.predict_output_word(context, topn=5)
    print("Top predictions:")
    for w,s in pred:
        print(f"{w}: {s:.2f}")
    print()
```

```python
for context in [['drive'],['drives'],['driving']]:
    print("Context:", context)
    pred = model.predict_output_word(context, topn=5)
    print("Top predictions:")
    for w,s in pred:
        print(f"{w}: {s:.2f}")
    print()
```

```python
for context in [['bank'],['rock']]:
    print("Context:", context)
    pred = model.predict_output_word(context, topn=5)
    print("Top predictions:")
    for w,s in pred:
        print(f"{w}: {s:.2f}")
    print()
```

```python
for context in [['bank', 'river'],['bank', 'finance']]:
    print("Context:", context)
    pred = model.predict_output_word(context, topn=5)
    print("Top predictions:")
    for w,s in pred:
        print(f"{w}: {s:.2f}")
    print()
```

From testing the similarity between semantically related words and different inflections of same lemma the answer seems to be that there is a higher similarity with semantically related words. The different inflections of the same lemma gave different nearest words where as the semantically related words gave similar nearest words.

Without context the ambiguous words gave nearest words that were associated with only one definition of the word. When adding context to the same word, then the nearest word was related to the ambiguous word in that context.
