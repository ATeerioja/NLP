# 2 A

All verb forms and their counts in descending order of frequency

1. Process data so that there is an array of every morphological feature
2. From that array count the number of occurances of each feature
3. Print the num of uniques and the top 10 most common forms

```python
from collections import Counter
```

```python
with open('fi_tdt-ud-train.conllu', mode='rt', encoding='utf-8') as f:
  data = f.read().split('\n')

data = [x for x in data if x.strip()]
```

```python
features = []

for row in data:
  if row[0] == "#":
    continue

  if row.split('\t')[5] == "_":
    continue

  features.append(row.split('\t')[5])

feature_counter = Counter(features)
```

```python
print(f'Number of unique forms of verbs: {len(feature_counter)}\n')
print(f'Top 10 most frequent forms of verbs: ')

for index, feature in enumerate(sorted(feature_counter.items(), key=lambda x: x[1], reverse=True)):
  print(f'{feature[0]} {feature[1]}')
  if index >= 10:
    break
```

# 2 B

1. Process the data so that each word is it's own row in an array
2. Count the number of unique words that are used as verbs and nouns
3. Print the number of pairs and some of the words

```python
with open('en_ewt-ud-train.conllu', mode='rt', encoding='utf-8') as f:
  data = f.read().split('\n')

data = [x for x in data if x.strip()]
```

```python
verbs = set()
nouns = set()

for row in data:
  if row[0] == "#":
    continue

  if row.split('\t')[3] == "NOUN":
    nouns.add(row.split('\t')[2])
  
  if row.split('\t')[3] == "VERB":
    verbs.add(row.split('\t')[2])
```

```python
verbnouns = set()

for noun in nouns:
  for verb in verbs:

    if noun == verb:
      verbnouns.add(noun)
```

```python
print(f'The number of unique verbs that are also used as nouns: {len(verbnouns)}\n')
print(f'Ten examples of verbs that are also used as nouns: ')

for index, word in enumerate(verbnouns):
  print(word)
  if index >= 10:
    break
```

3. print results

```python
with open('fi_tdt-ud-train.conllu', mode='rt', encoding='utf-8') as f:
  data = f.read().split('\n')

data = [x for x in data if x.strip()]
```

```python
features = set()
all_words = set()

for row in data:
  if row[0] == "#":
    continue

  if row.split('\t')[5] == "_":
    continue

  features.add(' ' .join(row.split('\t')[2:6]))
  all_words.add(row)
```

```python
print(f'Number of non ambiguous words: {len(features)}')
print(f'Number of words in total: {len(all_words)}')
print(f'The ambiguousness score of Finnish is: {100*(1 - (len(features)/len(all_words))):.1f}%')
```

The ambiguousness score tells the percentage of all words that have more than one meaning.
