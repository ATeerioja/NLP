### 1. Select a text classification corpus
https://huggingface.co/datasets/stanfordnlp/imdb

I chose the IMDB dataset as my text classification corpus. 
Only the strongly positive or strongly negative reviews are labelled, which makes the classification task challenging.
- It contains 50000 polarized movie reviews scraped from IMDB.
- It is split into train and test parts.
- Validation needs to be split from train.
- No more than 30 reviews per movie.

### 2. Interpret the dataset
This dataset has become a popular benchmark used to evaluate landmark models.

Splits:
- Train
- Test
- Unsupervised

Reviews:
- 25000 balanced pos / neg
- 25000 balanced pos / neg
- 50000 unlabelled

Fields:
text: raw movie review (string)
label: 0 = negative, 1 = positive

### 3. Identify state-of-the-art performance

Modern transformer models reach 95-97% test accuracy on the binary task.
- Leaderboard: https://nlpprogress.com/english/sentiment_analysis.html
